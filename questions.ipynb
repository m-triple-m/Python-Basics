```python
fruits = ['apple', 'guava', 'blueberry', 'pineapple', 'banana']
```

```python
# convert all the elements to uppercase and create a new list
# filter fruits with length more than 5
```

```python
upperFruits = [ f.upper() for f in fruits ]
upperFruits
```

```python
[ f for f in fruits if len(f) >5 ]
```

```python
# create a list of tuples where fruit is the first element
# and its length is the second element
# [ ('Apple', 5), ('Guava', 5), ('Blueberry', 9) ]
```

```python
[ (f, len(f)) for f in fruits ]
```
